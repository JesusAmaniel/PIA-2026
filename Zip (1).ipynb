```python

nums = [1, 2, 3, 4, 5, 6]
letras = ("a","e","i","o","u", "z", "x", "y")
print(nums)
print(letras)
```

```python
comb = zip(nums, letras)
print(f"El tipo del obj combinado resultante es {type(comb)}")
```

```python
rlista = list(comb)
print(f"El tipo del resultado combinado resultante es {rlista}")
```
