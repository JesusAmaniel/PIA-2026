```python
datos = np.random.randint(1,200, size = (18))
print(datos)

minimo = np.min(datos)
maximo = np.max(datos)
print(f"\nEl valor minimo es: {minimo}")
print(f"\nEl valor maximo es: {maximo}")

primer_cuartil = np.percentile(datos, 25)
mediana = np.percentile(datos, 50)
tercer_cuartil = np.percentile(datos, 75)

print(f"\nEl primer cuartil es: {primer_cuartil}")
print(f"\nLa mediana es: {mediana}")
print(f"\nEl tercer cuartil es: {tercer_cuartil}")
```

```python
import numpy as np

temperaturas = np.array([18,21,25,19,23,27,20])
promedio = np.mean(temperaturas)
dias_encima = np.sum(temperaturas > promedio)
farenheit = (temperaturas * 9/5) + 32

print(f"\nEl promedio de las temperaturas es: {promedio}")
print(f"\nDias con temperatura por encima del promedio: {dias_encima}")
print(f"\nLas temperaturas en grados Farenheit son: {farenheit}")
```
