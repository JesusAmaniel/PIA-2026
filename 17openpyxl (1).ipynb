```python
import openpyxl as op

libro = op.Workbook()
hoja = libro.active
hoja.title = "Alumnos"
print("Arhivo de Excel creado")

hoja["A1"] = "Nombre"
hoja["B1"] = "Edad"
hoja["C1"] = "Carrera"

hoja["A2"] = "Ana"
hoja["B2"] = 20
hoja["C2"] = "Administracion"
hoja["A3"] = "Pedro"
hoja["B3"] = 21
hoja["C3"] = "Limpieza"
hoja["A4"] = "Juan"
hoja["B4"] = 40
hoja["C4"] = "Contador"
hoja["A5"] = "Pablo"
hoja["B5"] = 34
hoja["C5"] = "Administracion"
```

```python
libro.save("alumnos.xlsx")
print("Archivo guardado correctamente")
```

```python
libro.close()
print("Archivo cerrado")
```

```python
libro = op.load_workbook("alumnos.xlsx")
hoja = libro["Alumnos"]
print("Archivo abierto correctamente")
```

```python
for fila in hoja.iter_rows(values_only=True):
    print(fila)
```
