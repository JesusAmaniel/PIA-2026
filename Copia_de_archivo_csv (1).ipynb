```python
import csv

data = dict()

auto1 = ("BMW", 2025)
auto2 = ("Toyota", 2015)

data[10] = auto1
data[20] = auto2

print(data)

filee = open("autos.csv", "w", encoding="latin1", newline="")

rec = csv.writer(filee)

rec.writerow(("Clave", "Marca", "Año"))

for clave, datos in data.items():
  rec.writerow((clave, datos[0], datos[1]))

filee.close()
```

```python
import csv

data = dict()

with open("autos.csv", "r", encoding="latin1", newline="") as filee:
  w = csv.reader(filee)
  next(w)

  for clave, marca, año in w:
    data[int(clave)] = (marca, int(año))

print(data)
```

```python
Entradas = [
    ['correo', 'nombre', 'telefono'],
    ['juan@gmail.com', 'Juan', '7468954554'],
    ['angei@gmail.com', 'Angel', '6845193421'],
    ['salamalenco@gmail.com', 'Brayan', '1245789856'],
    ['amanielel@gmail.com', 'Amaniel', '1765234789']
]

for e in Entradas:
    print(f'{e[0]}|{e[1]}|{e[2]}')
```

```python
print(">>> Contenido de la lista\n")
print(Entradas)

if os.path.exists("agenda.csv"):
  if (os.path.exists("agenda.bak")):
    os.remove("agenda.bak")
    os.rename("agenda.csv", "agenda.bak")
  else:
    os.rename("agenda.csv", "agenda.bak")

with open("agenda.csv", "w+") as f:
  for e in Entradas:
    f.write(f"{e[0]}|{e[1]}|{e[2]}\n")

print("\n>>> Contenido del archivo\n")
with open("agenda.csv") as f:
  print(f.read())
```
