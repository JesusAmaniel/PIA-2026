```python
#Copiado de listas
lista_base = ["a", "b", "c", "d"]
copia = lista_base #Esto NO copia la lista, genera un nuevo indentificador para el mismo objeto

#Formas de SI copiar una lista
copia_2 = lista_base.copy()
copia_3 = list(lista_base)
copia_4 = lista_base[:]

print(copia)
print(copia_2)
print(copia_3)
print(copia_4)
```

```python

lista_uno = [10, 20, 30, 40, 50]

#Escenario 1 - Método procedural imperativo
lista_resultante = []
for valor in lista_uno:
  lista_resultante.append(valor * 2)
print(f"resultado = {lista_resultante}")
```

```python
lista_uno = [10, 20, 30, 40, 50]

#Escenario 2 - Método declarativo (Comprensión de listas)
lista_con_resultado = [valor * 2 for valor in lista_uno]
print(f"resultado = {lista_con_resultado}")
```

```python
#Podemos obtener listas que contengan secuencias de valores
secuencia_entera = [valor for valor in range(1,101)]
print(secuencia_entera)
```

```python
lista_base = ["a", "b", 1, "c", 2, 3, "5"]

lista_destino = [valor for valor in lista_base if(isinstance(valor, str))]
print(lista_destino)
```
