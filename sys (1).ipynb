```python

import os, sys

l = list()
for num in range(1,1001):
  l.append(num)

consumoL = sys.getsizeof(l)
print(f"La lista tiene {len(l)} elementos, y tiene un tamaño de {consumoL} bytes")

t = tuple(l)
consumoT = sys.getsizeof(t)
print(f"La tupla tiene {len(t)} elementos, y tiene un tamaño de {consumoT} bytes")

proporcion = (consumoT/consumoL) * 100
print(f"La tupla consume {proporcion:.2f}% comparado a la lista")
```
