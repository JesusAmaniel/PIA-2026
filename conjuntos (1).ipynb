```python
#Creación de un conjunto con valores iniciales
mi_conjunto = {"A", "B", "C", "D"}

#Creación de un conjunto vacío
conjunto_vacio = set()

#Creación de un conjunto a partir de un iterable ya existente
lista_base = [10, 20, 30, 40, 20]
conjunto_numeros = set(lista_base)
print(conjunto_numeros)
```

```python
conjunto_uno = {"Hugo", "Paco", "Luis"}
conjunto_dos= {"Britni", "Llojan", "Luis"}
print("CONJUNTOS")
print(conjunto_uno)
print(conjunto_dos)
print()

#Todos los elementos de ambos conjuntos (UNIÓN)
print("UNIÓN")
print(conjunto_uno | conjunto_dos)
print()

#Aquellos elementos que existen TANTO EN UNO COMO EN OTRO conjunto (INTERSECCIÓN)
print("INTERSECCIÓN")
print(conjunto_uno & conjunto_dos)
print()

#Aquellos elementos en conjunto_uno que NO existen en conjunto_dos (DIFERENCIA)
print("DIFERENCIA")
print(conjunto_uno - conjunto_dos)
print()

#Aquellos elementos que estan en solamente uno de los dos conjuntos, pero no en ambos (DIFERENCIA SIMÉTRICA)
print("DIFERENCIA SIMËTRICA")
print(conjunto_uno ^ conjunto_dos)
```

```python
conjunto_tres = {"Paco", "Luis"}

#Consulta si el conjunto_tres es un subconjunto del conjunto_uno
print(conjunto_tres <= conjunto_uno)
```

```python

import random

valores = [random.randrange(1, 6) for valor in range(100)]
print(valores)

#Al obtener un conjunto a partir de la lista de valores aleatorios podremos identificar los valores únicos en ella
diferentes = set(valores)
print()
print(diferentes)
```

```python

conjunto = set()

conjunto.add(1)
print(conjunto)

conjunto.update([20, 30, 40])
print(conjunto)
```

```python
conjunto.remove(20)
print(conjunto)

conjunto.discard(30)
print(conjunto)
```
