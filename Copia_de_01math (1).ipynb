```python
import math


valorflontante = float(input("introducce un valor con fraccion decimal:\n"))

print(f"El siguiente entero hacia abajo de {valorflontante} es {math.floor(valorflontante)}")

print(f"El siguiente entero hacia arriba de {valorflontante} es {math.ceil(valorflontante)}")

print(f"La parte entera trucada de {valorflontante} es {math.math.trunc(valorflontante)}")#Equivale a flox
```

```python
valorNumerico = int(input ("\nDame un valor entero:\n"))

print(f"La raiz cuadrada de {valorNumerico} es igual a {math.sqrt(valorNumerico)}")

print(f"El factorial de {valorNumerico} es igual a {math.factorial(valorNumerico)}")
```
